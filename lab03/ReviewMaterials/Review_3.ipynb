## CodeBuilder!
This lab is extremely dense in its vocabulary and tools, and you really cannot use it to the full extent without knowing what to use where. I'll include a master list of all the table methods you have encountered, but **it's your job to pick amongst them**.

Because there is so much stuff, you can minimize sections within the codebuilder, and also double-click on a block to add it to the first empty slot.

### Library

This first table contains those methods that involve **reading/compiling/adding information into a table**.

|Purpose|Function|
|--|--|
|Creating a New Table (from arrays)|`NewTable = Table().with_columns(ColumnHeader1, ListOrArray1, ColumnHeader2, ListOrArray2, [etc, etc])`|
|Creating a New Table (from .csv file)|`NewTable = Table.read_table('FileName')`|
|Adding one row to a Table|`ExistingTable = ExistingTable.with_row(NewRowAsList)`|
|Adding multiple rows to a Table|`ExistingTable = ExistingTable.with_rows(listOfLists)`|
|Adding column(s) to a Table|`ExistingTable = ExistingTable.with_columns(ColumnHeader1, ListOrArray1, ColumnHeader2, ListOrArray2, [etc, etc])`|

This second table contains those methods that involve **modifying the appearance of a table**.

|Purpose|Function|
|--|--|
|Sorting the rows (ascending)|`ExistingTable.sort(ColumnHeader)`|
|Sorting the rows (descending)|`ExistingTable.sort(ColumnHeader, descending = True)`|
|Selecting column(s) in Table format|`ExistingTable.select(ColumnHeader1, ColumnHeader2, [etc])`|
|Getting rid of certain columns|`ExistingTable.drop(UnwantedColumnHeader1, UnwantedColumnHeader2, [etc])`|
|Restricting rows to certain parameters|`ExistingTable.where(ColumnHeader, predicate)`|
|Selecting any row(s)|`ExistingTable.take(IndexOfRow1, IndexOfRow2, [etc])`|
|Selecting the first n rows|`ExistingTable.take(np.arange(n))`|

Finally, this third table contains those methods that involve **getting information out of a table**.

|Purpose|Function|
|--|--|
|Extracting a column as an array|`ExistingTable.column(ColumnHeader)` OR `existingTable[ColumnHeader]`|
|Getting the number of rows/columns|`ExistingTable.num_rows` OR `ExistingTable.num_columns`|

Typically, datascience tables are very flexible: wherever you see `ColumnHeader`, you can replace that with the Index of that column. (first column is index 0, second is index 1, and so forth.)
