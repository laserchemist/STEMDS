## 1. Introduction to Tables

An **array** holds one attribute for many things.  
A **Table** holds *many* attributes — rows are entities, columns are measurements.

NASA's Exoplanet Archive catalogs planets discovered beyond our solar system.  
Below we build a table from two arrays describing confirmed exoplanets.

```python
# Confirmed exoplanets — a sample of well-characterised planets
planet_names = make_array(
    'Kepler-442b', 'Proxima Cen b', 'TRAPPIST-1e', 'K2-18b',
    'Kepler-62f',  '55 Cnc e',      'HD 40307g',   'GJ 667Cc',
    'tau Cet e',   'Kepler-452b')

# Distance from Earth in light-years
distances_ly = make_array(1206, 4.2, 39.5, 124, 1200, 41, 42, 23.6, 11.9, 1400)

exoplanets = Table().with_columns(
    "Planet",           planet_names,
    "Distance (ly)",    distances_ly
)   # column label      and the array of data
exoplanets
```

This builds a new table manually; you can add as many columns as you want, but you need to keep the pattern of alternating column label and array. All arrays must be the same size.

## 5. Filtering with `where`
The method `.where` is a very important function in this course. See below for its information.
<div style="background:#c1fcc0;border-left:6px solid #90EE90;padding:12px 16px;border-radius:6px;margin:10px 0;">
<b>New Method!</b> <code><b>existingTable</b>.where(<b>columnLabel</b>, <b>predicate</b>)</code><br>
This allows you to filter Table rows based on a specific column. That column's label is its first argument, while the second argument can be any number of predicates (listed below).
</div>


| Predicate | Meaning |
|-----------|---------|
| `are.equal_to(v)` | == v |
| `are.above(v)` | > v |
| `are.below(v)` | < v |
| `are.between(a, b)` | a ≤ value < b |
| `are.not_equal_to(v)` | ≠ v |

In this cell below, the `.where` method is used on the `spotify` table to get only the rows whose `Energy` `are.above` `80`.

```python
bangers = spotify.where('Energy', are.above(80))
bangers
```

### Adding computed columns
The extent of our ability as coders to analyze data is limited by what data is in the table itself. We can expand this ability by transforming existing columns and adding them to the table.

```python
# Add a "streams per BPM" column as a fun engagement metric
streams = spotify.column('Streams (millions)')
bpm     = spotify.column('BPM')
spotify = spotify.with_column('Streams/BPM', np.round(streams / bpm, 2))
spotify.sort('Streams/BPM', descending=True).select(['Track','Artist','Streams/BPM'])
```
