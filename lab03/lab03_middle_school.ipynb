## ⭐ Level 1 — What Is a Table?

An **array** stores one kind of information about many things.  
A **Table** stores *many kinds* of information — like a spreadsheet!

| | Array | Table |
|-|-------|-------|
| Shape | One column | Many columns |
| Example | `[38, 120, 70]` speeds | Speed + Name + Lifespan |

Let's build a table of amazing animals! 🦁

````python
# Two arrays — one attribute each
animal_names  = make_array('Cheetah','Peregrine Falcon','Sailfish',
                           'Pronghorn','Ostrich','Greyhound','Lion')
top_speeds_kph = make_array(120, 389, 110, 98, 70, 72, 80)

# Combine into a Table — rows stay together!
animals = Table().with_columns(
    "Animal",        animal_names,
    "Top Speed kph", top_speeds_kph
)
animals
````

<div style="background:#c1fcc0;border-left:6px solid #90EE90;padding:12px 16px;border-radius:6px;margin:10px 0;">
<b>New Method!</b> <code><b>newTable</b> = Table().with_columns(<b>Label1</b>, <b>Array1</b>, <b>Label2</b>, <b>Array2</b>, <b>[etc, etc]</b>)</code><br>
This builds a new table manually and stores it in <b>newTable</b>; you can add as many columns as you want, but you need to keep the pattern of alternating column label and array. All arrays must be the same size.
</div>

````python
# Sort to find the fastest!
animals.sort('Top Speed kph', descending=True)
````

<div style="background:#c1fcc0;border-left:6px solid #90EE90;padding:12px 16px;border-radius:6px;margin:10px 0;">
<b>New Method!</b> <code><b>existingTable</b>.sort(<b>columnLabel</b>)</code><br>
This sorts the rows according to the selected column label. You can also use the index of the column instead of the label.
</div>

## ⭐⭐⭐⭐ Level 5 — Filter with `where`

<div style="background:#c1fcc0;border-left:6px solid #90EE90;padding:12px 16px;border-radius:6px;margin:10px 0;">
<b>New Method!</b> <code><b>existingTable</b>.where(<b>columnLabel</b>, <b>predicate</b>)</code><br>
This allows you to filter Table rows based on a specific column. That column's label is its first argument, while the second argument can be any number of predicates (listed below).
</div>

| Predicate | Means | Example |
|-----------|-------|---------|
| `are.equal_to(v)` | equals v | `are.equal_to("cat")` |
| `are.above(v)` | greater than v | `are.above(100)` |
| `are.below(v)` | less than v | `are.below(50)` |
| `are.above_or_equal_to(v)` | greater than or equal to v | `are.above_or_equal_to(100)` |
| `are.below_or_equal_to(v)` | less than or equal to v | `are.below_or_equal_to(100)` |
| `are.between(a,b)` | a ≤ value < b | `are.between(50, 100)` |

````python
# Animals faster than 100 kph
speedsters = animals.where('Top Speed kph', are.above(100))
speedsters
````
