## ⭐ Level 1 — What Is a Table?

An **array** stores one kind of information about many things.  
A **Table** stores *many kinds* of information — like a spreadsheet!

| | Array | Table |
|-|-------|-------|
| Shape | One column | Many columns |
| Example | `[38, 120, 70]` speeds | Speed + Name + Lifespan |

Let's build a table of amazing animals! 🦁

````python
# Two arrays — one attribute each
animal_names  = make_array('Cheetah','Peregrine Falcon','Sailfish',
                           'Pronghorn','Ostrich','Greyhound','Lion')
top_speeds_kph = make_array(120, 389, 110, 98, 70, 72, 80)

# Combine into a Table — rows stay together!
animals = Table().with_columns(
    "Animal",        animal_names,
    "Top Speed kph", top_speeds_kph
)
animals
````

````python
# Sort to find the fastest!
animals.sort('Top Speed kph', descending=True)
````

### Adding rows to a Table

- **`.with_row([v1, v2, v3])`** — adds one row

````python
# Example: add one row
example = Table().with_columns("Number",[1,2], "Name",["One","Two"])
example = example.with_row([3, "Three"])
example
````

## ⭐⭐⭐⭐ Level 5 — Filter with `where`

`tbl.where("Column", are.predicate(value))` keeps only rows that match!

| Predicate | Means | Example |
|-----------|-------|---------|
| `are.equal_to(v)` | equals v | `are.equal_to("cat")` |
| `are.above(v)` | greater than v | `are.above(100)` |
| `are.below(v)` | less than v | `are.below(50)` |
| `are.between(a,b)` | a ≤ value < b | `are.between(50, 100)` |

````python
# Animals faster than 100 kph
speedsters = animals.where('Top Speed kph', are.above(100))
speedsters
````
