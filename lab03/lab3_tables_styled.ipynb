## 1. Introduction to Tables

An **array** describes one attribute of many things.  
A **Table** describes *many* attributes of many things — rows are entities, columns are attributes.

The cell below loads world population estimates (US Census Bureau, 1950–2015) into two arrays,  
then combines them into a Table so rows stay aligned.

```python
population_amounts = Table.read_table("world_population.csv").column("Population")
years = np.arange(1950, 2016, 1)
print("Population column:", population_amounts)
print("Years column:     ", years)
```

```python
population = Table().with_columns(
    "Year",       years,
    "Population", population_amounts
)
population
```

```python
population.sort('Population', descending=True)
```

## 5. Finding Pieces of a Dataset — `where`

`tbl.where("Col", are.predicate(value))` filters rows that satisfy a condition.

| Predicate | Meaning |
|-----------|---------|
| `are.equal_to(v)` | exactly v |
| `are.above(v)` | > v |
| `are.below(v)` | < v |
| `are.above_or_equal_to(v)` | ≥ v |
| `are.between(a, b)` | a ≤ value < b |
| `are.not_equal_to(v)` | ≠ v |

```python
imdb = Table.read_table('imdb.csv')
forties = imdb.where('Decade', are.equal_to(1940))
forties
```

| `tbl.take(rows)` | **Table** | you want specific row indices |

## Reference — Table Methods

<div style="background:#f0fff0;border-left:6px solid #27ae60;padding:10px 14px;border-radius:6px;margin:8px 0;font-size:0.95em;">

| Method | Example | Returns |
|--------|---------|---------|
| `Table()` | `Table()` | Empty table |
| `Table.read_table` | `Table.read_table("f.csv")` | Table from CSV |
| `with_columns` | `tbl.with_columns("A", arr)` | Table with new columns |
| `with_row` | `tbl.with_row([v1, v2])` | Table + one new row |
| `with_rows` | `tbl.with_rows([[…],[…]])` | Table + many new rows |
| `column` | `tbl.column("A")` | **Array** |
| `select` | `tbl.select("A","B")` | **Table** (subset of columns) |
| `drop` | `tbl.drop("A")` | **Table** (without listed columns) |
| `sort` | `tbl.sort("A", descending=True)` | Sorted Table copy |
| `where` | `tbl.where("A", are.above(5))` | Filtered Table copy |
| `take` | `tbl.take(np.arange(5))` | Table with those row indices |
| `num_rows` | `tbl.num_rows` | int |
| `num_columns` | `tbl.num_columns` | int |
