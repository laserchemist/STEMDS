### What the simulation tracks

| Metric | What it means |
|--------|--------------|
| 🌡️ Temperature | Average surface temperature in °C |
| 💧 Water Phase | Whether water is frozen, liquid, or vapor |
| 🌫️ CO₂ | How much carbon dioxide is in the atmosphere (greenhouse gas!) |
| 🧬 Life Score | How suitable conditions are for water-based life (0–100%) |

---

### The sliders explained

| Slider | What it controls | Real example |
|--------|-----------------|--------------|
| 🌟 Distance (AU) | How far from the star | Earth = 1.0, Mars = 1.52 |
| ☀️ Star Brightness | How bright/hot the star is | Sun = 1.0 |
| 🌫️ Starting CO₂ (%) | How much greenhouse gas at the start | Earth = 0.04%, Venus = 96% |
| 💧 Starting Water (%) | Water vapor in starting atmosphere | Earth-like = 1% |
| 🪨 Planet Mass | Size and gravity | Earth = 1.0, Mars = 0.11 |

## 🌌 Extension: Compare Two Planets

Run the cell below to see two planets side by side.  
Change the values in `planet_A` and `planet_B` to compare any two worlds!


```python
# ── Side-by-Side Comparison ───────────────────────────────────
from planet_sim import simulate_planet, plot_results, WATER_LABELS, WATER_EMOJI
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
%matplotlib inline

# Change these to compare any two planets!
planet_A = dict(dist=1.0,  lum=1.0, co2_0= 5.0, h2o_0=1.0, mass=1.00,
                label='Earth-like')
planet_B = dict(dist=1.52, lum=1.0, co2_0= 0.1, h2o_0=0.0, mass=0.11,
                label='Mars-like')

DARK_BG   = '#0f0f23'
PANEL_BG  = '#1a1a2e'
COLORS    = ['#74b9ff', '#fd79a8']

fig, axes = plt.subplots(2, 2, figsize=(13, 7), facecolor=DARK_BG)
fig.suptitle(f'Side-by-Side: {planet_A["label"]}  vs  {planet_B["label"]}',
             color='white', fontsize=13, fontweight='bold', y=0.98)

labels_done = False
for col, (p, color) in enumerate(zip([planet_A, planet_B], COLORS)):
    r = simulate_planet(p['dist'], p['lum'], p['co2_0'], p['h2o_0'], p['mass'])
    times = r['times']

    # Row 0: Temperature
    ax = axes[0][col]
    ax.set_facecolor(PANEL_BG)
    ax.axhspan(-20, 60, alpha=0.15, color='#55efc4')
    ax.plot(times, r['T_celsius'], color=color, lw=2.5)
    ax.set_title(f'{p["label"]} - Temperature', color='white', fontsize=10, fontweight='bold')
    ax.set_ylabel('Temp (deg C)', color='white')
    ax.tick_params(colors='white')
    ax.grid(color='#2a2a4e', lw=0.6)
    for sp in ax.spines.values(): sp.set_edgecolor('#444466')
    ax.text(0.03, 0.72, 'Life zone', transform=ax.transAxes,
             color='#55efc4', fontsize=8, style='italic')

    # Row 1: Habitability
    ax = axes[1][col]
    ax.set_facecolor(PANEL_BG)
    ax.fill_between(times, r['habitability'], alpha=0.4, color=color)
    ax.plot(times, r['habitability'], color=color, lw=2.5)
    ax.set_title(f'{p["label"]} - Life Score', color='white', fontsize=10, fontweight='bold')
    ax.set_ylabel('Life Score (0-1)', color='white')
    ax.set_xlabel('Time (million years)', color='white')
    ax.set_ylim(0, 1.05)
    ax.tick_params(colors='white')
    ax.grid(color='#2a2a4e', lw=0.6)
    for sp in ax.spines.values(): sp.set_edgecolor('#444466')

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```
