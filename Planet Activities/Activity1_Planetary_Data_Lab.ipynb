## Part 3 — Statistics

Now let's compute some **univariate statistics** on the full dataset.


```python
# ── Summary Statistics ────────────────────────────────────────────────────────

for col in ['Distance (AU)', 'Mass (Earth=1)', 'Radius (Earth=1)', 'Avg Temp (°C)']:
    data = planets.column(col)
    print(f"\n--- {col} ---")
    print(f"  Mean   : {np.mean(data):.2f}")
    print(f"  Median : {np.median(data):.2f}")
    print(f"  Min    : {np.min(data):.2f}  ({planets.column('Name')[np.argmin(data)]})")
    print(f"  Max    : {np.max(data):.2f}  ({planets.column('Name')[np.argmax(data)]})")
    print(f"  Std Dev: {np.std(data):.2f}")
```

```python
# ── Compare Solar System vs Exoplanets ────────────────────────────────────────

ss  = planets.where('Type', 'Solar System')
exo = planets.where('Type', 'Exoplanet')

print("Average Temperature (°C):")
print(f"  Solar System planets : {np.mean(ss.column('Avg Temp (°C)')):.1f} °C")
print(f"  Exoplanets           : {np.mean(exo.column('Avg Temp (°C)')):.1f} °C")

print("\nAverage Mass (Earth = 1):")
print(f"  Solar System planets : {np.mean(ss.column('Mass (Earth=1)')):.1f}")
print(f"  Exoplanets           : {np.mean(exo.column('Mass (Earth=1)')):.1f}")

print("\n🔍 Why might these averages look so different?")
print("   Hint: think about which planets are in each group...")
```

## Part 4 — Visualizations

Graphs help us spot patterns that are hard to see in raw numbers.


```python
# ── Plot 1: Temperature Bar Chart ─────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(14, 5))

all_names  = list(planets.column('Name'))
all_temps  = list(planets.column('Avg Temp (°C)'))
all_types  = list(planets.column('Type'))

bar_colors = ['steelblue' if t == 'Solar System' else 'tomato' for t in all_types]

bars = ax.bar(all_names, all_temps, color=bar_colors, edgecolor='white', linewidth=0.6)

ax.axhline(0,   color='black', linewidth=0.8, linestyle='--', alpha=0.5, label='Freezing (0 °C)')
ax.axhline(-20, color='cyan',  linewidth=0.8, linestyle=':',  alpha=0.7, label='Habitability lower bound (−20 °C)')
ax.axhline(60,  color='orange',linewidth=0.8, linestyle=':',  alpha=0.7, label='Habitability upper bound (60 °C)')
ax.axhspan(-20, 60, alpha=0.08, color='green')

ax.set_ylabel('Average Temperature (°C)', fontsize=12)
ax.set_title('Surface Temperature of All 22 Worlds', fontsize=14, fontweight='bold')
ax.set_xticks(range(len(all_names)))
ax.set_xticklabels(all_names, rotation=45, ha='right', fontsize=8)

ss_patch  = mpatches.Patch(color='steelblue', label='Solar System')
exo_patch = mpatches.Patch(color='tomato',    label='Exoplanet')
ax.legend(handles=[ss_patch, exo_patch], loc='lower right', fontsize=9)
plt.tight_layout()
plt.show()
print("\n🔍 Which exoplanets fall within the green habitable temperature band?")
```

```python
# ── Plot 2: Temperature Histogram ─────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(9, 5))

ss_temps  = ss.column('Avg Temp (°C)')
exo_temps = exo.column('Avg Temp (°C)')

ax.hist(ss_temps,  bins=10, alpha=0.65, color='steelblue', label='Solar System', edgecolor='white')
ax.hist(exo_temps, bins=10, alpha=0.65, color='tomato',    label='Exoplanets',   edgecolor='white')

ax.axvspan(-20, 60, alpha=0.12, color='green', label='Habitable range')
ax.set_xlabel('Average Temperature (°C)', fontsize=12)
ax.set_ylabel('Number of Planets', fontsize=12)
ax.set_title('Temperature Distribution: Solar System vs Exoplanets', fontsize=13, fontweight='bold')
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# ── Plot 3: Mass vs Radius (log–log) ─────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 6))

masses    = planets.column('Mass (Earth=1)')
radii     = planets.column('Radius (Earth=1)')
types_col = planets.column('Type')
rocky_col = planets.column('Rocky?')

for i in range(len(all_names)):
    c = 'steelblue' if types_col[i] == 'Solar System' else 'tomato'
    m = 'o' if rocky_col[i] else 's'
    ax.scatter(masses[i], radii[i], c=c, marker=m, s=130,
               edgecolors='white', linewidth=0.7, zorder=3)
    ax.annotate(all_names[i], (masses[i], radii[i]),
                textcoords='offset points', xytext=(5, 3), fontsize=7)

ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlabel('Mass (Earths)', fontsize=12)
ax.set_ylabel('Radius (Earths)', fontsize=12)
ax.set_title('Planet Mass vs Radius', fontsize=13, fontweight='bold')
ax.grid(True, alpha=0.3)

# Highlight Earth
earth_idx = all_names.index('Earth')
ax.scatter(masses[earth_idx], radii[earth_idx],
           c='limegreen', marker='*', s=300, zorder=5, label='Earth ⭐')

ss_patch  = mpatches.Patch(color='steelblue', label='Solar System')
exo_patch = mpatches.Patch(color='tomato',    label='Exoplanet')
circle    = plt.Line2D([0],[0], marker='o', color='w', markerfacecolor='gray',
                        markersize=9, label='Rocky')
square    = plt.Line2D([0],[0], marker='s', color='w', markerfacecolor='gray',
                        markersize=9, label='Gas/Ice')
ax.legend(handles=[ss_patch, exo_patch, circle, square], fontsize=9)
plt.tight_layout()
plt.show()
print("\n🔍 What pattern do you see for rocky planets vs gas giants?")
```

```python
# ── Plot 4: The Goldilocks Zone ───────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(14, 5))

temps_all = planets.column('Avg Temp (°C)')
norm      = plt.Normalize(vmin=temps_all.min(), vmax=temps_all.max())
cmap      = plt.cm.RdYlBu_r

y_ss = np.zeros(ss.num_rows)
y_ex = np.ones(exo.num_rows)

sc1 = ax.scatter(ss.column('Distance (AU)'), y_ss,
                 c=ss.column('Avg Temp (°C)'), cmap=cmap, norm=norm,
                 s=200, edgecolors='white', linewidth=1, zorder=3)
sc2 = ax.scatter(exo.column('Distance (AU)'), y_ex,
                 c=exo.column('Avg Temp (°C)'), cmap=cmap, norm=norm,
                 s=200, edgecolors='white', linewidth=1, marker='^', zorder=3)

for row in ss.rows:
    ax.annotate(row[0], (row[2], 0), textcoords='offset points',
                xytext=(0, -18), ha='center', fontsize=8)
for row in exo.rows:
    ax.annotate(row[0], (row[2], 1), textcoords='offset points',
                xytext=(0, 10), ha='center', fontsize=7)

# Goldilocks zone (approximate, Sun-like star)
ax.axvspan(0.85, 1.70, alpha=0.18, color='green', label='Goldilocks Zone (Sun-like)')

ax.set_xlim(0.01, 40)
ax.set_xscale('log')
ax.set_yticks([0, 1])
ax.set_yticklabels(['Solar System', 'Exoplanets'], fontsize=11)
ax.set_xlabel('Distance from Star (AU)', fontsize=12)
ax.set_title('The Goldilocks Zone — Not Too Hot, Not Too Cold', fontsize=13, fontweight='bold')

cbar = plt.colorbar(sc1, ax=ax, pad=0.02)
cbar.set_label('Surface Temp (°C)', fontsize=10)
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()
print("\n🔍 Which exoplanets cluster closest to the Goldilocks Zone?")
```

## Part 5 — Habitability Score

Let's build a simple **Habitability Index** using three criteria:
1. **Temperature** — must be between −20 °C and 60 °C for liquid water
2. **Rocky** — gas giants can't support surface life as we know it
3. **Water** — presence or possibility of liquid water scores extra points


```python
# ── Compute Habitability Score (0–10) ─────────────────────────────────────────
# habitability_score() is defined in activity1_init.py

scores = []
for row in planets.rows:
    temp  = row[5]   # Avg Temp (°C)
    rocky = row[8]   # Rocky?
    water = row[6]   # Has Water?
    scores.append(habitability_score(temp, rocky, water))

planets = planets.with_column('Habitability (0-10)', scores)

# Sort by habitability
sorted_planets = planets.sort('Habitability (0-10)', descending=True)
sorted_planets.select('Name', 'Type', 'Avg Temp (°C)', 'Has Water?',
                       'Rocky?', 'Habitability (0-10)').show()
```

```python
# ── Habitability Bar Chart ────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(14, 5))

sorted_names  = list(sorted_planets.column('Name'))
sorted_scores = list(sorted_planets.column('Habitability (0-10)'))

bar_colors = []
for s in sorted_scores:
    if   s >= 7:  bar_colors.append('#2ecc71')   # green
    elif s >= 4:  bar_colors.append('#f39c12')   # orange
    elif s >= 1:  bar_colors.append('#e74c3c')   # red
    else:         bar_colors.append('#7f8c8d')   # grey

bars = ax.bar(sorted_names, sorted_scores, color=bar_colors,
              edgecolor='white', linewidth=0.8)

for bar, score in zip(bars, sorted_scores):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
            f'{score:.1f}', ha='center', va='bottom', fontsize=8, fontweight='bold')

ax.set_ylim(0, 11)
ax.set_ylabel('Habitability Score (0–10)', fontsize=12)
ax.set_title('Planetary Habitability Index — All 22 Worlds', fontsize=14, fontweight='bold')
ax.set_xticks(range(len(sorted_names)))
ax.set_xticklabels(sorted_names, rotation=40, ha='right', fontsize=8)

legend_patches = [
    mpatches.Patch(color='#2ecc71', label='Good (≥ 7)'),
    mpatches.Patch(color='#f39c12', label='Marginal (4–7)'),
    mpatches.Patch(color='#e74c3c', label='Poor (1–4)'),
    mpatches.Patch(color='#7f8c8d', label='Uninhabitable'),
]
ax.legend(handles=legend_patches, loc='upper right', fontsize=9)
plt.tight_layout()
plt.show()
```
