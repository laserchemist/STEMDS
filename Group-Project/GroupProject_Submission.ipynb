### 🚀 Step 3 — Submit!
<div style="background:#9E1B34;color:#fff;padding:10px 18px;border-radius:6px;font-size:14px;display:inline-block;margin-top:4px;">

```python
# Re-running this cell is safe — preserves the button and output widget.
if '_gp_submit_init' not in dir():
    btn_submit = widgets.Button(
        description='🚀 Submit Project Proposal',
        layout=widgets.Layout(width='270px', height='48px'),
        style={'button_color': '#9E1B34', 'font_weight': 'bold'}
    )
    out_submit = widgets.Output()
    _gp_submit_init = True

def on_submit(_):
    with out_submit:
        clear_output(wait=True)

        # --- Validate ---
        team        = w_team.value.strip()
        title       = w_title.value.strip()
        dataset     = w_dataset.value
        hypothesis  = w_hypothesis.value.strip()
        description = w_description.value.strip()

        errors = []
        if not team:               errors.append('⚠️ Group Name is required')
        if not title:              errors.append('⚠️ Project Title is required')
        if dataset == DATASETS[0]: errors.append('⚠️ Please select a dataset')
        if not hypothesis:         errors.append('⚠️ Hypothesis is required')
        if not description:        errors.append('⚠️ Project Description is required')

        members = [
            {'name': nw.value.strip(), 'role': rw.value if rw.value != '---' else 'Unassigned'}
            for nw, rw in member_widgets if nw.value.strip()
        ]
        if not members: errors.append('⚠️ At least one team member is required')

        if errors:
            display(HTML('<div style="color:#9E1B34;font-weight:bold;font-size:13px;">' + '<br>'.join(errors) + '</div>'))
            return

        # --- Build entry ---
        entry = {
            'team': team, 'title': title, 'dataset': dataset,
            'hypothesis': hypothesis, 'description': description,
            'members': members,
            'timestamp': time.strftime('%Y-%m-%d %H:%M:%S'),
            'submitted_at': time.strftime('%H:%M')
        }

        # --- Atomic write ---
        try:
            with open(SUBMISSIONS_FILE) as f:
                subs = json.load(f)
        except (FileNotFoundError, json.JSONDecodeError):
            subs = []
        subs = [s for s in subs if s.get('team', '').lower() != team.lower()]
        subs.append(entry)
        dir_ = os.path.dirname(SUBMISSIONS_FILE)
        os.makedirs(dir_, exist_ok=True)
        with tempfile.NamedTemporaryFile('w', dir=dir_, delete=False, suffix='.tmp') as f:
            json.dump(subs, f, indent=2)
            tmp = f.name
        os.replace(tmp, SUBMISSIONS_FILE)

        # --- Confirmation card ---
        member_rows = ''.join(
            f'<tr><td style="padding:3px 8px;color:#555;">{m["name"]}</td>'
            f'<td style="padding:3px 8px;color:#9E1B34;font-weight:500;">{m["role"]}</td></tr>'
            for m in members
        )
        display(HTML(f'''
        <div style="background:#F5F2EC;border:2px solid #9E1B34;border-radius:10px;
                    padding:20px 24px;max-width:580px;font-family:system-ui,sans-serif;">
          <div style="font-size:1.4em;">✅ <strong style="color:#9E1B34;">Proposal Submitted!</strong></div>
          <div style="margin:8px 0 4px;font-size:12px;color:#888;">Submitted at {entry["timestamp"]}</div>
          <table style="width:100%;border-collapse:collapse;margin-top:10px;">
            <tr><td style="padding:4px 0;font-weight:600;color:#444;width:130px;">Group:</td>
                <td style="color:#1A1226;font-weight:700;font-size:1.05em;">{team}</td></tr>
            <tr><td style="padding:4px 0;font-weight:600;color:#444;">Title:</td>
                <td style="color:#333;font-style:italic;">{title}</td></tr>
            <tr><td style="padding:4px 0;font-weight:600;color:#444;">Dataset:</td>
                <td style="color:#333;">{dataset}</td></tr>
          </table>
          <div style="margin-top:12px;font-weight:600;color:#444;font-size:13px;">Team Members:</div>
          <table style="margin-top:4px;">{member_rows}</table>
          <div style="margin-top:12px;background:#fff;border-radius:6px;padding:10px 12px;">
            <div style="font-weight:600;color:#9E1B34;font-size:12px;">HYPOTHESIS</div>
            <div style="font-size:13px;color:#333;margin-top:4px;font-style:italic;">{hypothesis}</div>
          </div>
          <div style="margin-top:10px;font-size:12px;color:#888;">
            You can re-submit at any time to update your proposal — it replaces your previous entry.
          </div>
        </div>
        '''))

# Clear any previously registered handlers before re-registering (safe on re-run)
btn_submit._click_handlers.callbacks.clear()
btn_submit.on_click(on_submit)
display(btn_submit, out_submit)
```
