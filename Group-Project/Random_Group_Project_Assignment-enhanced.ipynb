```python
def make_random_groups(people, number_of_groups,labels):
    
    #Shuffle list of people
    random.shuffle(people)
    
    #Create groups
    all_groups = []
    for index in range(number_of_groups):
        group = people[index::number_of_groups]
        all_groups.append(group)
    
    #Format and display groups
    print("✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨")
    print("✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨")
    sleep(5)
    for index, group in enumerate(all_groups):
        print(f"✨{labels[index]}✨: {' / '.join(group)}\n")
        sleep(2)
        print("✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨✨")
        sleep(1)
    return all_groups
```

```python
from datetime import datetime, timedelta
import json, random as _r
from IPython.display import HTML, display

# ── Customize ─────────────────────────────────────────────────────
CLASS_START_TIME   = "14:00"   # 24-hour HH:MM
CLASS_DURATION_MIN = 80
SETUP_MIN          = 5
WRAP_UP_MIN        = 5
PRESENTATION_MIN   = None      # None = auto-divide equally
COURSE_NAME        = "SCTC 1013 – Elements of Data Science"
# ──────────────────────────────────────────────────────────────────

try:
    _num          = len(GROUP)
    _group_labels = list(Groups[:_num])
except NameError:
    _group_labels = ["Hacking Hooters","Data Owls","Broad Street Bootstrappers",
                     "Data Smiths","Norwegian Numpys","Stellas Statisticians","Plotting Phils"]
    _num = len(_group_labels)

_avail    = CLASS_DURATION_MIN - SETUP_MIN - WRAP_UP_MIN
_slot_min = PRESENTATION_MIN if PRESENTATION_MIN else (_avail // _num)

_base_dt = datetime.strptime(CLASS_START_TIME, "%H:%M") + timedelta(minutes=SETUP_MIN)
_slots = []
for i in range(_num):
    s = _base_dt + timedelta(minutes=i * _slot_min)
    e = s + timedelta(minutes=_slot_min)
    _slots.append(
        f"{s.strftime('%I:%M').lstrip('0')}–{e.strftime('%I:%M %p').lstrip('0')}"
    )

print(f"\U0001f4cb  {_num} groups  \u00d7  {_slot_min} min each  =  {_num * _slot_min} min presentations")
print(f"    Setup: {SETUP_MIN} min  |  Wrap-up: {WRAP_UP_MIN} min  |  Total: {CLASS_DURATION_MIN} min")
print()
print(f"  {'#':>2}   {'Time slot':<22}  Group label")
print("  " + "\u2500" * 54)
for i, (sl, gl) in enumerate(zip(_slots, _group_labels)):
    print(f"  {i+1:>2}.  {sl:<22}  {gl}")
```

```python
_SEG_COLORS = ['#D85A30','#1D9E75','#378ADD','#BA7517',
               '#993556','#534AB7','#639922','#A32D2D']
_seg_colors = (_SEG_COLORS * 4)[:_num]

# ── Unique suffix prevents getElementById finding a stale element
#    when the cell is re-run in the same JupyterLab session
_uid = _r.randint(100000, 999999)

_js_data = (
    json.dumps({
        'groups':  _group_labels,
        'slots':   _slots,
        'colors':  _seg_colors,
        'slotMin': _slot_min,
        'course':  COURSE_NAME,
        'uid':     str(_uid),
    })
    .replace('<', r'\u003c')
    .replace('>', r'\u003e')
    .replace('&', r'\u0026')
)

# IDs are parameterised so the template itself is uid-agnostic
_WHEEL_HTML = r"""
<div id="pw-root-%%UID%%" style="font-family:'Segoe UI',Arial,sans-serif;max-width:780px;border:1px solid #e2e2e2;border-radius:12px;overflow:hidden;margin:8px 0;">

  <div style="background:#f8f8f8;padding:12px 20px;border-bottom:1px solid #e8e8e8;display:flex;align-items:center;gap:10px;">
    <div style="font-size:20px;">&#127905;</div>
    <div>
      <div style="font-size:15px;font-weight:600;color:#222;">Presentation Slot Spinner</div>
      <div id="pw-course-%%UID%%" style="font-size:12px;color:#888;"></div>
    </div>
  </div>

  <div style="display:flex;background:#fff;">

    <div style="padding:20px;flex:0 0 auto;text-align:center;border-right:1px solid #f2f2f2;">

      <div style="background:#f7f7f7;border:1px solid #e8e8e8;border-radius:8px;padding:10px 18px;margin-bottom:14px;min-width:380px;">
        <div style="font-size:10px;color:#bbb;text-transform:uppercase;letter-spacing:2px;margin-bottom:3px;">now spinning for</div>
        <div id="pw-team-%%UID%%" style="font-size:17px;font-weight:600;color:#222;">&#8212;</div>
      </div>

      <div style="position:relative;display:inline-block;margin-bottom:12px;">
        <div style="position:absolute;top:-14px;left:50%;transform:translateX(-50%);font-size:20px;color:#222;z-index:5;line-height:1;">&#9660;</div>
        <canvas id="pw-canvas-%%UID%%" width="420" height="420"
          style="display:block;border-radius:50%;border:1px solid #ddd;"></canvas>
      </div>

      <div>
        <button id="pw-btn-%%UID%%"
          style="background:#222;color:#fff;border:none;padding:10px 36px;font-size:15px;font-weight:600;border-radius:24px;cursor:pointer;transition:opacity 0.2s;">
          Spin!
        </button>
      </div>
      <div id="pw-status-%%UID%%" style="font-size:12px;color:#888;margin-top:8px;min-height:16px;"></div>
    </div>

    <div style="padding:20px;flex:1;min-width:0;">
      <div style="font-size:10px;font-weight:600;color:#bbb;text-transform:uppercase;letter-spacing:2px;padding-bottom:8px;border-bottom:1px solid #f0f0f0;margin-bottom:10px;">Class Schedule</div>
      <div id="pw-list-%%UID%%" style="display:flex;flex-direction:column;gap:5px;"></div>
      <div id="pw-export-%%UID%%" style="display:none;margin-top:14px;">
        <button id="pw-exportBtn-%%UID%%"
          style="background:#27ae60;color:#fff;border:none;padding:10px 16px;border-radius:8px;cursor:pointer;font-size:13px;font-weight:600;width:100%;">
          &#8595; Download animated schedule page (HTML)
        </button>
      </div>
    </div>

  </div>
</div>

<script>
/* ── guard: defer until DOM is ready, retry if element not yet painted ── */
(function () {
  var DATA = %%DATA%%;
  var UID  = DATA.uid;

  function init() {
    var canvas = document.getElementById('pw-canvas-' + UID);
    if (!canvas) { setTimeout(init, 80); return; }

    var groups  = DATA.groups, slots = DATA.slots, colors = DATA.colors;
    var slotMin = DATA.slotMin, course = DATA.course;
    var n       = groups.length, TWO_PI = Math.PI * 2, slice = TWO_PI / n;

    var angle = 0, spinning = false, currentIdx = 0;
    var usedSlots   = new Set();
    var assignments = {};

    var ctx     = canvas.getContext('2d');
    var cx = canvas.width / 2, cy = canvas.height / 2;
    var R  = Math.min(cx, cy) - 6;

    var teamEl   = document.getElementById('pw-team-'   + UID);
    var statusEl = document.getElementById('pw-status-' + UID);
    var listEl   = document.getElementById('pw-list-'   + UID);
    var exportEl = document.getElementById('pw-export-' + UID);
    var spinBtn  = document.getElementById('pw-btn-'    + UID);

    document.getElementById('pw-course-' + UID).textContent = course;
    teamEl.textContent = groups[0];
    drawWheel(0);
    renderList();

    spinBtn.addEventListener('click', spin);
    spinBtn.addEventListener('mouseenter', function () { if (!spinning) this.style.opacity = '0.75'; });
    spinBtn.addEventListener('mouseleave', function () { this.style.opacity = '1'; });

    /* ── draw ──────────────────────────────────────────────────── */
    function drawWheel(a) {
      ctx.clearRect(0, 0, canvas.width, canvas.height);
      for (var i = 0; i < n; i++) {
        var s = -Math.PI / 2 + a + i * slice, e = s + slice;
        var taken = usedSlots.has(i);
        ctx.beginPath(); ctx.moveTo(cx, cy);
        ctx.arc(cx, cy, R, s, e); ctx.closePath();
        ctx.fillStyle = taken ? '#d5d5d5' : colors[i];
        ctx.fill(); ctx.strokeStyle = '#fff'; ctx.lineWidth = 2.5; ctx.stroke();
        ctx.save(); ctx.translate(cx, cy); ctx.rotate(s + slice / 2);
        ctx.textAlign = 'right';
        ctx.fillStyle = taken ? '#aaa' : '#fff';
        ctx.font = '600 ' + (n <= 6 ? 16 : 13) + 'px "Segoe UI",Arial,sans-serif';
        ctx.fillText(slots[i], R - 9, 4);
        ctx.restore();
      }
      ctx.beginPath(); ctx.arc(cx, cy, 16, 0, TWO_PI);
      ctx.fillStyle = '#fff'; ctx.fill();
      ctx.strokeStyle = '#ccc'; ctx.lineWidth = 1; ctx.stroke();
      ctx.fillStyle = '#ccc'; ctx.font = '11px serif';
      ctx.textAlign = 'center'; ctx.textBaseline = 'middle';
      ctx.fillText('\u2605', cx, cy);
    }

    /* ── spin ──────────────────────────────────────────────────── */
    function spin() {
      if (spinning || currentIdx >= n) return;
      var avail = [];
      for (var i = 0; i < n; i++) { if (!usedSlots.has(i)) avail.push(i); }
      if (!avail.length) return;

      var target     = avail[Math.floor(Math.random() * avail.length)];
      var baseAngle  = -((target + 0.5) * slice);
      var tAngle     = baseAngle;
      while (tAngle <= angle) tAngle += TWO_PI;
      tAngle += (6 + Math.floor(Math.random() * 4)) * TWO_PI;
      tAngle += (Math.random() - 0.5) * slice * 0.4;

      spinning = true;
      spinBtn.disabled = true; spinBtn.style.opacity = '0.4';
      statusEl.textContent = 'Spinning\u2026';

      var t0 = angle, t1 = tAngle;
      var dur = 3500 + Math.random() * 700, tStart = performance.now();

      function frame(now) {
        var p = Math.min((now - tStart) / dur, 1);
        angle = t0 + (t1 - t0) * (1 - Math.pow(1 - p, 5));
        drawWheel(angle);
        if (p < 1) { requestAnimationFrame(frame); return; }

        angle = t1; drawWheel(angle); spinning = false;
        usedSlots.add(target);
        assignments[groups[currentIdx]] = { slot: slots[target], color: colors[target] };
        statusEl.textContent = groups[currentIdx] + '  \u2192  ' + slots[target];
        currentIdx++; renderList();

        if (currentIdx < n) {
          teamEl.textContent = groups[currentIdx];
          setTimeout(function () {
            spinBtn.disabled = false; spinBtn.style.opacity = '1';
            statusEl.textContent = '';
          }, 900);
        } else {
          teamEl.textContent = 'All groups assigned!';
          spinBtn.style.display = 'none';
          exportEl.style.display = 'block';
        }
      }
      requestAnimationFrame(frame);
    }

    /* ── live schedule panel ───────────────────────────────────── */
    function renderList() {
      listEl.innerHTML = '';
      groups.forEach(function (g, i) {
        var a = assignments[g], isNext = (i === currentIdx);
        var row = document.createElement('div');
        row.style.cssText =
          'display:flex;justify-content:space-between;align-items:center;' +
          'padding:7px 10px;border-radius:6px;background:' + (a ? a.color + '0d' : 'transparent') + ';' +
          'border:1px solid ' + (a ? a.color + '33' : isNext ? '#ccc' : '#f2f2f2') + ';' +
          'border-left:3px solid ' + (a ? a.color : isNext ? '#555' : '#e8e8e8') + ';';
        var nameD = document.createElement('div');
        nameD.style.cssText = 'font-size:13px;font-weight:' + (a || isNext ? '600' : '400') +
          ';color:' + (a ? '#222' : isNext ? '#333' : '#bbb') + ';';
        nameD.textContent = (isNext ? '\u25b6  ' : '') + g;
        var slotD = document.createElement('div');
        slotD.style.cssText = 'font-size:12px;font-weight:600;white-space:nowrap;' +
          'color:' + (a ? a.color : isNext ? '#aaa' : '#ddd') + ';';
        slotD.textContent = a ? a.slot : (isNext ? 'spinning next\u2026' : 'pending');
        row.appendChild(nameD); row.appendChild(slotD); listEl.appendChild(row);
      });
    }

    /* ── build & download animated HTML schedule page ─────────── */
    document.getElementById('pw-exportBtn-' + UID).addEventListener('click', function () {

      /* sort by chronological slot order */
      var sorted = Object.keys(assignments).sort(function (a, b) {
        return slots.indexOf(assignments[a].slot) - slots.indexOf(assignments[b].slot);
      });

      /* CSS keyframe slide-up cards, staggered per group */
      var cardCSS = sorted.map(function (g, i) {
        return '.card-' + i + '{animation:rise 0.55s ease both;animation-delay:' + (i * 0.18) + 's}';
      }).join('\n');

      /* timeline bar segments */
      var barSegs = sorted.map(function (g) {
        var a = assignments[g];
        return '<div style="flex:1;background:' + a.color +
               ';display:flex;align-items:center;justify-content:center;' +
               'font-size:10px;color:#fff;font-weight:600;padding:0 4px;white-space:nowrap;overflow:hidden;text-overflow:ellipsis;">' +
               g + '</div>';
      }).join('<div style="width:2px;background:#fff;flex:0 0 auto;"></div>');

      /* group cards */
      var cards = sorted.map(function (g, i) {
        var a = assignments[g];
        return '<div class="card card-' + i + '" style="border-left:5px solid ' + a.color + ';">' +
          '<div class="card-time">' + a.slot + '</div>' +
          '<div class="card-name">' + g + '</div>' +
          '</div>';
      }).join('');

      var html = '<!DOCTYPE html>\n<html lang="en">\n<head>' +
        '<meta charset="UTF-8">' +
        '<meta name="viewport" content="width=device-width,initial-scale=1">' +
        '<title>Presentation Schedule \u2013 ' + course + '</title>\n' +
        '<style>\n' +
        '@keyframes rise{from{opacity:0;transform:translateY(22px)}to{opacity:1;transform:none}}\n' +
        'body{font-family:"Segoe UI",Arial,sans-serif;background:#f4f5f8;margin:0;padding:32px 24px;}\n' +
        'h1{margin:0 0 4px;font-size:22px;font-weight:700;color:#1a1a1a;}\n' +
        '.sub{font-size:14px;color:#888;margin:0 0 28px;}\n' +
        '.timeline{display:flex;height:38px;border-radius:8px;overflow:hidden;margin-bottom:32px;border:1px solid #ddd;}\n' +
        '.cards{display:grid;grid-template-columns:repeat(auto-fill,minmax(220px,1fr));gap:14px;max-width:860px;}\n' +
        '.card{background:#fff;border-radius:10px;padding:16px 18px;border:1px solid #ebebeb;' +
             'box-shadow:0 2px 6px rgba(0,0,0,0.06);transition:box-shadow 0.2s;cursor:default;}\n' +
        '.card:hover{box-shadow:0 6px 18px rgba(0,0,0,0.12);}\n' +
        '.card-time{font-size:12px;font-weight:600;color:#aaa;letter-spacing:0.5px;margin-bottom:6px;}\n' +
        '.card-name{font-size:17px;font-weight:700;color:#222;}\n' +
        '.foot{margin-top:36px;font-size:12px;color:#bbb;text-align:center;}\n' +
        cardCSS + '\n' +
        '</style></head>\n<body>\n' +
        '<h1>Presentation schedule</h1>\n' +
        '<p class="sub">' + course + ' &nbsp;&middot;&nbsp; ' + slotMin + '\u202fmin per group</p>\n' +
        '<div class="timeline">' + barSegs + '</div>\n' +
        '<div class="cards">' + cards + '</div>\n' +
        '<p class="foot">Generated with Presentation Slot Spinner</p>\n' +
        '</body></html>';

      var link = document.createElement('a');
      link.href = URL.createObjectURL(new Blob([html], { type: 'text/html' }));
      link.download = 'presentation_schedule.html';
      link.click();
    });

  } /* end init() */

  setTimeout(init, 50);

})();
</script>
"""

_html_out = (_WHEEL_HTML
    .replace('%%UID%%',  str(_uid))
    .replace('%%DATA%%', _js_data))

display(HTML(_html_out))
```
