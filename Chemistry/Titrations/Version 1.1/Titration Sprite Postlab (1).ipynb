## Step 3: The Equivalence Points
Oh BOY! I love paragraphs and I know you do too. :)

The behavior of acids and bases is very very complicated, and if we tried to explain it all to you it would melt your brains faster than, well, acid. For the purposes of this lab, all you'll need to know is that the pH of a solution depends on how much acid and base is in there, and how strong the acid/base is. Stronger/more acid correspond to a **lower pH**, and stronger/more base correspond to a **higher pH**. 

The strength of an acid (its acidity) depends on how many hydrogens it still has. Thus, $C_6H_8O_7$ (what we call the triprotic acid) is much more acidic than $C_6H_7O_7^-$ (diprotic), which is more acidic than $C_6H_6O_7^{2-}$ (monoprotic). Thus, as you steady deprotonate the acid by adding base, the pH of the solution will go from extremely acidic to just slightly acidic, and eventually neutral and then basic.

Recall back to the Copper Lab. When the reactant amounts of that lab matched the ratios on the reaction formula, both of them reacted in such a way that there was no reactant left behind; you (hopefully) should have seen that that trial had the most product. In this titration case, this idea of reaction amounts will come into play again, but in a different form. In **any** of the three reactions above, when the amount of base matches the amount of acid, the pH becomes very sensitive; just a very small amount of base will trigger massive changes in pH. You'll see this as the vertical jumps in pH that appear on your graph, and we call the center of those jumps the **equivalence point**.
