## Slope

The best fit line is in the format
$$y = mx + b$$
where
* x is the variable of the x-axis
* y is the variable of the y-axis
* m is the slope (that is, how much does a change in x change y?)
* b is the intercept (that is, if x were 0, what would y be?)

While there are many ways to get the best fit, a common way (and the way that Excel and others do it) is with something called the **Least Squares Method**.

We can get m through the following formula:

$$m = \frac{n\Sigma(x_iy_i)-\Sigma(x_i)\Sigma(y_i)}{n\Sigma(x^2_i)-(\Sigma(x_i))^2}$$

where 
* $x_i$ and $y_i$ represent the x value and y value, respectively, at some position i
* $\Sigma$ is just a fancy way of writing sum. So, $\Sigma(x_i)$ means that you are adding up all the x values.
* $n$ is the number of terms.

## The R² Value
When we analyze a best fit line, a chief concern is how accurate this best fit line is in respect to the data. If you see that the points all lie very near to the line, then the best fit line is quite accurate.

A more mathematical way to analyze this is to calculate the R² value.
$$R^2 = 1 - \frac{\Sigma((y_{i} - \hat{y}_{i})^2)}{\Sigma((y_{i} - \bar{y})^2)}$$
where
* $y_i$ is the actual y (in the data)
* $\hat{y}_i$ is the predicted $y_i$ based on the best fit line and and the corresponding $x_i$
* $\bar{y}$ is the mean of all $y_i$

In the numerator, we are...
* subtracting each y by their predicted value
* squaring these differences
* and then summing these squared differences up

In the denominator, we are...
* subtracting each y by the mean of y
* squaring these differences
* and then summing these squared differences up
