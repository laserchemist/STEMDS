The reaction you conducted earlier is what's called a double displacement reaction. This is of the form

$$\textcolor{red}{A}\textcolor{blue}{C} + \textcolor{red}{B}\textcolor{blue}{D} \rightarrow \textcolor{red}{A}\textcolor{blue}{D} + \textcolor{red}{B}\textcolor{blue}{C}$$

where $\textcolor{red}{A}$ and $\textcolor{red}{B}$ are the positively-charged cations and $\textcolor{blue}{C}$ and $\textcolor{blue}{D}$ are the negatively-charged anions.
