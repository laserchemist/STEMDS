## Reaction 4: Acid-Base Neutralization
In chemistry, an acid often refers to a molecule that can donate a Hydrogen. Conversely, a base refers to a molecule that can accept a Hydrogen. Here, Hydrochloric Acid ($HCl$) is the acid, and Sodium Hydroxide ($NaOH$) is the base.

When an acid and base mix, the Hydrogen from the acid attaches to the base. In response, the Sodium ($Na$) will detach and move to the Chlorine ($Cl$). This compound, $NaCl$, is what we often call table salt.

```python
reactionbuilder(4, RB_WIDGETS)
```
