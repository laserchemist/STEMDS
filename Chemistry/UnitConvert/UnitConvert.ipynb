Converting between units is one of the most common applications of math in the sciences. This page is designed to help make this process easy, understandable, and comfortable.

The most straightforward way to approach this is to imagine each conversion as multiplying the value by a fraction whose amount is equal to 1. 

For example, consider the following: $$30\; hours\;=\; ?\; seconds$$

You know the folloing conversions:

* 1 hour = 60 minutes
* 1 minute = 60 seconds

Take that first conversion as an example: since one hour is exactly the same time as 60 minutes, then you can be sure that the fraction $\frac{60 minutes}{1 hour}$ is equal to 1.

So, when you multiply $30 hours$ by $\frac{60 minutes}{1 hour}$, the $hour$ units (in parentheses) cancel out, and what you are left with is the quantity in minutes:

$$30\; (hours)\;*\; \frac{60\; minutes}{1\;(hour)} =\; 30*60\; minutes\;=1800\; minutes$$

This can be strung together across multiple unit conversions, with the only unit not cancelled out being the unit of the result. Always make sure that you are properly cancelling out the units (here shown in parentheses); that is your signal that you are converting correctly.

$$30\; (hours)\;*\; \frac{60\; (minutes)}{1\;(hour)}\;*\; \frac{60\; seconds}{1\;(minute)} =\; 30*60*60\; seconds=10800\; seconds$$
