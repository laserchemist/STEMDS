## Guide to Code Structures
We've often seen that the biggest hurdle to learning to code is the sheer complexity of it all. One of the best ways to overcome that is to realize that, from labs 0 to 3, all of the code that you will do and be expected to know fall into one of four code structures.

### Operation
**General Structure:** `a <operation> b`

This is the most familiar, as its format almost exactly matches that of lementary math operations. These are most directly studied in Labs 00 and 01. 

A related cousin to operations is the **conditional** (in Lab 05), which also has parallels to math, but only outputs a True or False.

```python
# examples

op1 = 1 + 2

op2 = 3 * 6

op2 % op1 # modulus operator: the remainder when op2 is divided by op1
```

```python
'water' + 'bottle'
```

```python
# conditional example

# you can think of this as a statement:
# 'one is less than 2'
# and python will tell you whether that is true or false.
1 < 2
```

### Function
**General Structure:** `functionName(a, b)`

The Function is used extensively from as early as Lab 01. Most of these will be imported from libraries like numpy and matplotlib, so they will often have `np.` or `plt.` before their function name. Don't confuse them with methods though!

```python
abs(-3)
```

```python
print(3, 5, 7)
```

```python
import numpy as np
np.arange(10)
```

### Method
**General Structure:** `object.methodName(a, b)`

The Method is very similar to the function, with one major difference: In a method, the first argument (the object) is located *before* the method name, and the method acts on this object. How it do so is governed by the other arguments. This is entirely separate from the library notation (e.g. `np.arange`) that you'll see on functions.

You'll see this used quite sparingly, mostly for lists, tables, and strings (Labs 02 and 03).

```python
numbers = [5, 7, 9]

numbers.append(3) # acts on the list called numbers!

numbers
```

```python
word = 'python'

word.split('y') # acts on the string called word!
```

### Index
**General Structure:** `a[b]`

This is by far the most limited in its purpose, as it exclusively refers to position. That is, in a list/array/string called `a`, that general structure will output the element at position `b`. This is very similar to the notation of lists; the main way to differentiate them is that indices will never use commas within the brackets.

```python
numbers = [1, 4, 6, 9, 10]

numbers[4] # gets term at 4th position (positions start at 0)
```

```python
word = 'soup'

word[0]
```
