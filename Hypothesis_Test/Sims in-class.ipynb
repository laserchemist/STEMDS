# Simulation In-Class Demonstration

```python
from datascience import *
import matplotlib.pyplot as plt
%matplotlib inline
plt.style.use('ggplot')
import numpy as np
import random
from potatoes import *
```

```python
og, sims = sheets_import(True)
```

```python
og
```

```python
sim1 = sims.where(0, 1)
sim2 = sims.where(0, 2)
sim3 = sims.where(0, 3)
sim4 = sims.where(0, 4)
```

## The Actual Test Statistic

```python
north = og.where('Location', 'North')
south = og.where('Location', 'South')
test_stat = np.mean(south.column('Temperature')) - np.mean(north.column('Temperature'))
test_stat
```

## Simulations Results

```python
sims_results = []
for i in [sim1, sim2, sim3, sim4]:
    north_sim = i.where('Simulated Location', 'North')
    south_sim = i.where('Simulated Location', 'South')
    test_stat_sim = np.mean(south_sim.column('Temperature')) - np.mean(north_sim.column('Temperature'))
    sims_results.append(test_stat_sim)
```

```python
plt.hist(test_stat_sim, color='blue', label='simulated test stats')
plt.scatter(test_stat, 0, s=300, color='red', label='actual test statistic', marker='^')
plt.legend()
plt.show()
```
