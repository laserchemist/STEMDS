# P-tests for other kinds of data
If you were observant, you may have noticed that the kind of p-test above relies on one of the variables consisting of just two values (here, location consists of just north and south). This is a **discrete** variable, in contrast with the more **continuous** temperature variable. But what if both values are continuous? Or both discrete?

Comfortingly, all you need to change is the way you get the actual and random test statistics. All other parts are the exact same.
* If both variables are continuous and made of numbers, the test statistic is the slope of the best fit line between them. (see code below)
* If both are discrete, the test statistic is just the difference between the number of observations in each
    * (e.g. if testing port vs north/south, the test statistic is the number of northern port cities minus number of southern port cities)
    * The groundhog test from lab 5 is a slightly more complex version that makes use of the unique circumstances surrounding groundhog predictions. <br>(the two variables are spring onset and seeing shadow)

You'll notice that we had asked you for your city's population (a continuous variable) and whether it is a port city (a discrete variable). Conduct the following p-tests:
* Location vs Population (one of each kind of variable; same process as above)
* Population vs Temperature (two continuous variables)
* Location vs Port city (two discrete variables)

```python
# use this to get the slope between any two numerical columns in any table t

def slope(t, label_x, label_y):
    x = t.column(label_x)
    y = t.column(label_y)
    standard_x = (x - np.mean(x))/np.std(x)
    standard_y = (y - np.mean(y))/np.std(y)
    r = np.mean(standard_x * standard_y)
    return r*np.std(y)/np.std(x)
```
