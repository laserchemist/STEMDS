### Hypothesis Framework

| | Statement |
|---|---|
| **Null Hypothesis H₀** | Groundhog predictions are no better than random guessing (≈50%) |
| **Alternative Hypothesis H₁** | Groundhog predictions are better than random |

We can only *reject H₀* — we never "prove" H₁. The strength of evidence is the **p-value**:
the probability of getting a result at least as extreme as observed, *assuming H₀ is true*.

| p-value | Interpretation |
|---------|---------------|
| > 0.05 | Insufficient evidence to reject H₀ |
| ≤ 0.05 | Suggestive evidence against H₀ |
| ≤ 0.01 | Strong evidence against H₀ |

## ⚠️ The Multiple Comparisons Problem

We just ran **three separate hypothesis tests** on the same dataset:

| Test | Subject | Variable |
|------|---------|----------|
| 1 | All groundhogs combined | `p_All` |
| 2 | Essex Ed (EE) | `p_EE` |
| 3 | Punxsutawney Phil (PYPL) | `p_PYPL` |

This raises a serious statistical concern that practicing scientists must grapple with.

<div style="background:#fff0f0;border-left:6px solid #e74c3c;padding:12px 16px;border-radius:6px;margin:10px 0;">
<b>The Problem: Inflated False Positive Rate</b><br><br>
Each individual test has a 5% chance of producing a false positive —
incorrectly rejecting H₀ even when the groundhog has no real ability
— purely by luck.<br><br>
When you run <em>multiple</em> tests and report whichever ones look significant,
that false positive rate compounds. With 3 independent tests each at α = 0.05:<br><br>
<code>P(at least one false positive) = 1 − (1 − 0.05)³ = 1 − 0.95³ ≈ 0.143</code><br><br>
There is already a <b>14% chance</b> of at least one spurious significant result
— nearly triple the intended 5% — before we've even looked at the data.
Now imagine the researchers screened all 33 groundhogs individually...
</div>

````python
# Visualise how the false positive rate grows with number of tests
import numpy as np
import matplotlib.pyplot as plt

num_tests = np.arange(1, 34)   # 1 to 33 groundhogs
alpha     = 0.05

# P(at least one false positive) = 1 - P(none)
p_at_least_one_fp = 1 - (1 - alpha)**num_tests

plt.figure(figsize=(9, 4))
plt.plot(num_tests, p_at_least_one_fp, color='#e74c3c', linewidth=2.5)
plt.axhline(0.05, color='navy',   linestyle='--', label='Intended α = 0.05')
plt.axhline(0.50, color='grey',   linestyle=':',  label='50% false positive rate')
plt.axvline(3,  color='#c800a0', linestyle='--', alpha=0.7, label='Our 3 tests')
plt.axvline(33, color='#ff6600', linestyle='--', alpha=0.7, label='All 33 groundhogs')
plt.fill_between(num_tests, p_at_least_one_fp, alpha=0.12, color='#e74c3c')
plt.xlabel('Number of hypothesis tests performed')
plt.ylabel('P(at least one false positive)')
plt.title('The Multiple Comparisons Problem\n'
          'Testing more groundhogs → more likely to find a false "winner"')
plt.legend(loc='lower right')
plt.ylim(0, 1)
plt.tight_layout()
plt.savefig('multiple_comparisons.png', dpi=120)
plt.show()
print(f'With 3 tests:  P(≥1 false positive) = {1-(0.95**3):.3f}')
print(f'With 33 tests: P(≥1 false positive) = {1-(0.95**33):.3f}')
````

### The Bonferroni Correction

One standard remedy is the **Bonferroni correction**: divide your significance
threshold α by the number of tests performed.

**α_adjusted = α / n_tests**

<div style="background:#f0fff0;border-left:6px solid #27ae60;padding:12px 16px;border-radius:6px;margin:10px 0;">
<b>Applied to our groundhog study:</b><br><br>
If we test all 33 groundhogs individually plus the group as a whole
(34 tests total), the adjusted threshold becomes:<br><br>
<code>α_adjusted = 0.05 / 34 ≈ 0.0015</code><br><br>
A groundhog's p-value would need to be <em>below 0.0015</em> — not just 0.05 —
to be considered genuinely significant after accounting for the search.
</div>

````python
# How many tests did we effectively run?
# 33 individual groundhogs + 1 combined group = 34 potential tests.
n_tests          = 34
alpha_original   = 0.05
alpha_bonferroni = alpha_original / n_tests

print(f'Original threshold:   α = {alpha_original:.4f}')
print(f'Bonferroni-corrected: α = {alpha_bonferroni:.4f}  (= 0.05 / {n_tests})')
print()
print('How our p-values compare to the corrected threshold:')
for label, p in [('All groundhogs', p_All), ('Essex Ed', p_EE), ('Punxsutawney Phil', p_PYPL)]:
    verdict = '✅ Still significant' if p < alpha_bonferroni else '❌ No longer significant'
    print(f'  {label:22s}  p = {p:.4f}   {verdict}')
````
