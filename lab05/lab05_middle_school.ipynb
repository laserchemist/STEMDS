## ⭐⭐ Level 2 — Conditional Statements: Code That Decides

An `if/elif/else` block runs different code depending on what's true.

```python
if <condition>:
    <do this if True>
elif <other condition>:
    <do this instead>
else:
    <do this if nothing matched>
```

Only **one** branch ever runs — Python picks the first `True` one and skips the rest.

<div style="background:#c1fcc0;border-left:6px solid #90EE90;padding:12px 16px;border-radius:6px;margin:10px 0;">
<b>New Structure!</b> <code>if/elif/else</code> statements<br>
Study the above cell very carefully! <code>if/elif/else</code> statements are your primary tool for separating inputs. 
</div>

````python
# Example: letter grade function
def letter_grade(score):
    if score >= 90:
        # all scores 90 or above go down this path, and not any of the other paths.
        return 'A'
    elif score >= 80:
        # we do not have to specify that these scores are less than 90, because the first path already took them out.
        return 'B'
    elif score >= 70:
        return 'C'
    else:
        return 'F'

print(letter_grade(95))   # A
print(letter_grade(82))   # B
print(letter_grade(65))   # F
````

## ⭐⭐⭐ Level 4 — Probability: How Likely Is It?

**Probability** is a number between 0 and 1 that describes how likely something is.

| Probability | Meaning |
|-------------|---------|
| 0 | Impossible — will never happen |
| 0.5 | 50/50 — equally likely either way |
| 1 | Certain — will always happen |

**Key rule:** All probabilities for an event must add up to 1.  
So if something has probability `p`, then *not* that thing has probability `1 - p`.

**Independent events:** If two events don't affect each other,
multiply their probabilities: `P(A and B) = P(A) × P(B)`

````python
# Probability of rolling a 6 on a standard die
p_six = 1/6
print(f"P(rolling a 6)      = {p_six:.4f}")
print(f"P(NOT rolling a 6)  = {1 - p_six:.4f}")
print(f"P(two 6s in a row)  = {p_six * p_six:.4f}")
````

### Step 4 — The p-value: Putting a Number on "Is It Luck?" 🎯

The **p-value** answers: *"What fraction of random simulations did as well or better than the real groundhog?"*

- **p-value close to 0** → very few random simulations beat the groundhog → likely real ability!
- **p-value close to 0.5** → about half of random simulations beat the groundhog → looks like luck
- **Rule of thumb:** scientists usually say a p-value less than 0.05 (5%) is "worth paying attention to"
  and a p-value less than 0.01 (1%) is "strong evidence".<br>

Think of it like this: if you claimed your coin was special because it landed heads 53 times out of 100, but that happens randomly 25% of the time — that's not very convincing. But if something happens randomly only 1% of the time, that's much more surprising!

````python
# p-value: what fraction of random simulations matched or beat the real groundhogs?
p_All = np.count_nonzero(hogsimdata.column('Correct') >= percent_correct/100) / num_simulations
print(f"📊 All groundhogs p-value: {p_All:.3f}")
print()
if p_All < 0.01:
    print("🌟 p < 0.01 — Strong evidence groundhogs beat chance!")
elif p_All < 0.05:
    print("👀 p < 0.05 — Suggestive, worth investigating further")
else:
    print("🪙 p ≥ 0.05 — No evidence groundhogs beat a coin flip")
````
