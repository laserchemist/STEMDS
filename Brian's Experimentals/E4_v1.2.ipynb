## The Purpose of a *Custom* Function

A function's general purpose is to *do something* and spit out the end result in return. Python itself is built with a whole host of functions and methods, and the libraries we import carry with themselves functions of their own.

The purpose of a custom function is not to create something entirely different from these already-established functions. After all, you cannot make something new without building on what is already established. Instead, a custom function serves the main purpose of **Abbreviation**; it packs into a concise one-line format what was originally many lines of code. Think of it this way: you would never say 'light amplification by stimulated emission of radiation', you'd always say 'laser', even though the object that the two terms are referring to is exactly the same.

When do you want to make custom functions?
* **Repetition:** You anticipate that you'll be using a certain set of code lines over and over again.
* **Organization:** You have a set of code lines that serve a distinct purpose and want to distinguish them from other lines.
* **Variability:** You anticipate that you'll be using a certain set of code lines that largely change depending on the inputs.

## The Structure of a Function

A function consists of five parts:
* Function Name: This is self-explanatory; it is what the kernel identifies the function as (like the `arange` in `np.arange(start, stop, step)`)
* Arguments: In the `np.arange(start, stop, step)` example, these are the `start, stop, step`. They are what change from use to use and what allow your function to serve many different purposes.
* DocString: They are the first line after you define the function that tells you what the function's purpose is. They are surrounded in triple quotes.
* Code: These are the code within the function itself that transform the Arguments into the Output.
* Output: Marked by a **`return`**; they are what your function spits out and what other variables can be assigned as.

See the following for an example.

```python
def get_average(arrayOrList): # here, the function name is in Blue! Thank you, Jupyter! The arguments are unworthy of being blue.
    """A function that gets the average of the inputted Array or List""" # The DocString, summarizing what this function does
    summation = sum(arrayOrList) # these two lines of code use the argument (arrayOrList) and treat it like a variable in its own right.
    mean = summation/len(arrayOrList)
    return mean # this last line determines what the output is. You'll see what this means later on.
```

```python
def impotent_get_average(arrayOrList):
    """A function that gets the average of the inputted Array or List, but doesn't have a return line"""
    summation = sum(arrayOrList)
    mean = summation/len(arrayOrList)
    print(mean) # this lack of return means that it doesn't have any real use in regards to storing the mean as a variable.
```

Having a return line means that, in the above example, the mean can be stored in a variable and worked with in further code. If you do not make it return the mean, it is but a 'NoneType' husk and thus not really usable in further operations.

```python
mean1 = get_average([1,2,3,4,5])
```

```python
mean1
```

```python
mean2 = impotent_get_average([1,2,3,4,5]) # it is automatically printing the mean since we had commanded the function to do so.
# this 3.0 is NOT what mean2 is stored as. mean2 does not have anything but a NoneType.
```

```python
mean2 # nothing happened, because it is a NoneType.
```

```python
mean1+1
```

```python
mean2+1
```

## Matplotlib graphing
Matplotlib is the standard library we will use for graphing data. The action of importing matplotlib is a bit more complex than the single-line imports of other libraries like numpy and datascience. This is because matplotlib is extremely diverse in its use and appearance, and the additional lines clarify the exact formatting for matplotlib. You will need three lines:
* `%matplotlib inline`: This tells matplotlib to put the graphs in the output of jupyter cells instead of in a separate window.
* `import matplotlib.pyplot as plt`: This grabs from the overall library the group of functions/methods that are in the `pyplot` sublibrary. This is the most used and simplest version of matplotlib, and sufficient for most applications.
* `plt.style.use('ggplot')`: This tells matplotlib that the main style of graphing is what is defined as `ggplot`. It is named for Grammar of Graphics, which was adhered to in the creation of the graphing application in the R language. Other styles include `default`, `classic`, `fivethirtyeight`, and `dark_background` (but you can always search for more in the official documentation).

## Datascience mapping
If you have a dataset that incorporates latitude and longitude coordinates, you can plot it on a map! :D

Don't fret, it uses the `datascience`, so if you've done `from datascience import *`, the mapping should work fine as well. In fact, it accepts the data (y, x, labels, and more!) as columns in a singular table format! You do need to change the column headers to explicitly say what they represent, and y (latitude) and x (longitude) have to be the first two columns.

There are three main pieces of code that you should be aware of. I'm assigning them to variables `Points`, `Atlas`, and `Layer`, simply because having them as variables is pretty necessary if you want to use the map repeatedly and creatively. In reality, you can name them however.
* `Points = Circle.map_table(dataTable)`: The crux of mapmaking! This puts a `Circle` or `Marker` (adjust between them by replacing the Circle on the front) on the map for each of the locations in your dataset. As is, it has default settings on the appearance of these circles/markers. To manually change these settings, add them as additional arguments (e.g. `color = 'red'`). This singular function is enough for if you need to plot a singular dataset.
* `Atlas = Map(location = (ycenter, xcenter), zoom_start = defaultZoomAsNumber)`: Technically, you can do `Map()` without inputs, but at least specifying where the map is positioned when you first load it up is pretty handy for navigation. This includes a host of other functions pertaining to the map window, including height, width, background (aka `tiles`), all of which are adjusted through additional arguments.
* `Layer = Atlas.overlay(dataTable)`: Operated and customized very similarly to `map_table()`, but this is intended for adding additional datasets to an already existing map (hence why we use the `Atlas` variable here). Absolutely essential if you want to put multiple datasets on a single map.
