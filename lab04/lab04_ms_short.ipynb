## Part 1: Writing Functions
Here, you will learn how to write your own functions that you can use alongside those already in Python.

Before you can use your custom functions, you must define them. You can put a lot of complexity into a custom function, so we'll start simple and build up from there.

### Function 1: A super simple function. No parameters, no return value.
This is as simple a function as you can get. Note that most of the details of the function lie in the first line:
<ul>
<li>
It starts with a <b><code><font color='green'>def</font></code></b>. This is how you tell python that you're making a new function.</li>
<li>
Then, the function name. In this case, it is <code><font color='blue'>print_one_dad_joke</font></code>.</li>
<li>
After that, any arguments are indicated in parentheses. Here, there are no arguments.
</li>
<li>
And finally, a colon (<code>:</code>) to end the starting line.
</li>
</ul>

After the starting line is the <b>function body</b>, which is code that the function will carry out when it is called. In this example, all it will do is print out the joke. The function body can be as long as you want, but it must be indented; the first line without the indent marks the end of the function.

````python
def print_one_dad_joke():                                                       # the starting line
    print("What did one plant say to the other? Aloe! Long thyme no see.")      # the function body

# ^ note the gap (indent) between the left margin and the code line
````

Once we've defined it, we can use it as much as we want.

````python
print_one_dad_joke()
````

### Function 2: Add a function parameter.

You may realize that Function 1 is quite inflexible. Because it has no arguments, it prints the same joke each and every time you use it. Now, let's make a new version that can change its behavior.

Here, we will add an **argument** (<code>joke</code>) to the function. Whatever string is placed in the parentheses, the function will assign <code>joke</code> to it, and then print it out.

````python
def print_any_dad_joke(joke):
    print(joke)
````

````python
dad_joke1 = "I only seem to get sick on weekdays. I must have a weekend immune system"
dad_joke2 = "What brand of underwear do chemists wear? Kelvin Klein."

print_any_dad_joke(dad_joke1)       # joke = dad_joke1
print_any_dad_joke(dad_joke2)       # joke = dad_joke2
````

Now, you may be wondering: if I already have a variable <code>joke</code> outside the function, and I run <code>print_any_dad_joke</code>, will the <code>joke</code> in the function replace my old <code>joke</code>?

````python
joke = "I hate my job; all I do is crush cans all day. It’s soda pressing."

print('The variable joke before I run the function:', joke, '\n')

print_any_dad_joke("How do cows stay up to date? They read the Moo-spaper.")

print('\nThe variable joke after I run the function:', joke)
````

As you can see, running the function had no change on the variable `joke`. In fact, all variables you make inside a function will not exist outside the function.

### Function 3: Return a value from a function.
Now, there is one limitation in Functions 1 and 2 that we didn't address, and that is the idea of getting values from a function. If you try to assign a variable to Function 2, all you'll get is a NoneType.

````python
dad_joke = "Where do pirates get their hooks? Second hand stores."

printed_joke = print_any_dad_joke(dad_joke)            # the joke appears because we had commanded it to print

print('\nThe variable I got is', printed_joke)     # but we can't assign the printed joke to a variable
````

We solve this through a <code><font color='green'><b>return</b></font></code> line. This serves as the 'output' of the function, in that any value here will be assigned to a variable.

````python
def return_any_dad_joke(joke):
    
    return joke             # this outputs the joke instead of just printing it
````

````python
returned_joke = return_any_dad_joke(dad_joke)       # we can assign a variable to the function

print('The variable I got is', returned_joke)     # and it works!
````

### Function 4: A function that takes more than one input parameter.
Your function can be designed to accept more than one input parameter. Let's write a function that accepts two, yes two, dad jokes and returns the total number of characters in the two jokes combined.

````python
def total_joke_length(joke1, joke2):
    length_joke1 = len(joke1)
    length_joke2 = len(joke2)
    return length_joke1 + length_joke2
````

````python
dad_joke1 = "What do you call a beehive without an exit? Unbelievable."
dad_joke2 = "Did you know that the first french fries weren’t cooked in France? They were cooked in Greece."

total_joke_length(dad_joke1, dad_joke2)
````

There are a couple of things to notice here:
- First, as mentioned before, the variables inside the function have the names given in the function definition, not the names passed to the function.
- Second, we used the `return` statement to pass back total number of characters in the two jokes.
- Third, these jokes are killing me!

### Function 5: A function can return more than one value
Just as a function can be defined to accept multiple input parameters, functions can return multiple output parameters. Let's modify this last function to return the length of the both jokes rather than the total.

````python
def joke_length(joke1, joke2):
    length_joke1 = len(joke1)
    length_joke2 = len(joke2)
    return length_joke1, length_joke2
````

````python
dad_joke1 = "If prisoners could take their own mug shots…They’d be called cellfies."
dad_joke2 = "I just broke up with my mathematician girlfriend. She was obsessed with an X."

length1, length2 = joke_length(dad_joke1, dad_joke2)
print("Length of joke 1: ", length1)     # length1 is assigned to length_joke1
print("Length of joke 2: ", length2)     # length2 is assigned to length_joke2
````

The function returns two values, so when calling the function you can put two variable on the left of the `=` to receive the return values.

### Function 6: Keyword parameters
Sometimes you want a function that has options. You want to give the user choices in how to use the function without having to write multiple versions of the same function. This can be accomplished using `keyword parameters.` These parameters have default values that will be used if the user doesn't change them.

Let's write a function that takes a dad joke and returns the length of the joke, but gives the user the option of also printing the joke, or not.

````python
def joke_length_with_modifier(joke, modifier=0):
    
    # here, python will assume modifier to be 0 if you don't mention it

    # but if you assign modifier to be a value, python will obey
    
    return len(joke) + modifier
````

````python
dad_joke = "If a pig loses its voice…does it become disgruntled?"

joke_length_with_modifier(dad_joke)   # we didn't mention the modifier, so python assumes it's 0
````

````python
joke_length_with_modifier(dad_joke, modifier=3)   # now we do mention it, and so python changes it to 3
````

You might recognise that you have used keyword parameters before, such as when you used the Table sort method with `descending=True`, which changed the default sort behavior for a table.

### Putting it all together
Take a look at the function below. It does the following:
- Takes in the input parameters `a` and `b`.
- Also takes in the optional parameters `xlo` and `xhi` (they are assumed to be -10 and 10 unless changed)
- Generates a sequence of x-values, and defines y in terms of it
- Plots a line $ y = ax + b $

````python
def linear(a, b, xlo=-10, xhi=10):
    """Plot a line"""
    x = np.arange(xlo, xhi, 0.5)     # this generates the x values
    y = a * x + b                    # and this line gets the y values
    plt.plot(x, y, '-*')
    return y
````

````python
# Call the function accepting the default values for the keyword parameters
y = linear(2, 10)
````

````python
y = linear(2, 10, xlo=-15)
````

````python
# Call the function overriding the default values for xlo and xhi
y = linear(2, 10, xlo=-15, xhi=20)
````
