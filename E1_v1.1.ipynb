## The Syntax of Python
Python, like all computer languages, adheres to a very strict set of rules regarding how code is written and interpreted. All of the code you write will be a combination of four main code structures.

|Name|Structure|Example|Purpose|
|--|--|--|--|
|Operation|`object1 <operator> object2`|`a + b`|Create a new value that is a modification of a and b given the operator specified.|
|Function|`functionName(arguments)`|`print(a, b)`|Take in some inputs (called arguments) and do something with them as befit its design. Functions are by far the most versatile and most common of the four structures, so expect lots of variety.|
|Method|`object.methodName(arguments)`|`a.replace(b, c)`|A specialized version of the function; it modifies the object as befit its design AND any specifications described in the arguments.|
|Index|`object[position]`|`a[b]`|Spoiler for lab 02; this is for data types that are a sequence of characters (e.g. strings, lists, arrays). The index concerns the position of each term in the sequence.|

Note that in the wild, these structures are not limited to their forms as described in the table. You can have a method with no arguments, and indices that specify more than one position. However, even with such variations, these structures remain remarkably close to the purposes as specified here. Moreover, and this is VERY common, these structures will be nested within each other. For example, `functionName(a[b])`; in this case, the innermost (the index) is evaluated first, and that becomes the argument for the function.

As a baseline rule, do NOT mix up these structures with any variables you create. If you name a variable the exact same thing as a function or method name, python will become very confused and you will likely be unable to use that function/method again unless you restart the kernel.

## An intro to Importing Libraries

```python
import numpy as np
import math as m
from datascience import *
```

The above 'importing' of these libraries, as they are called, is akin to the unlocking of a whole new set of functions, methods, and other workings to be used in your code. If you do not include and run these imports, you won't have access to these tools.

These additional tools' appearances depend on how you import them. Specifically, their functions typically appear with the name of their library before it. For example, since we imported numpy as np, its functions have 'np' before it. See below for examples.

```python
np.arange(5)
```

```python
arange(5)
```

```python
import numpy
numpy.arange(5)
```

See how, since we imported numpy again but with no clarification on its name, we can thus use its full name 'numpy' before the function name. (Since we also imported it as 'np' earlier, the 'np.arange' format still works.)

This means you can technically use anything under the sun as the name of a library.

```python
import numpy as tomato
tomato.arange(5)
```

If you really REALLY want to import the library without its name, it requires a bit of formatting and the use of an asterisk. This is super risky, however, and should be used only if there are very few functions in the library (like in the Datascience module.)

```python
from numpy import *
arange(5)
```

This 'from (method) import (function)' can be used to render specific functions without their module name. If I run the cell below this without having done what I did in the previous cells, only arange would be able to be used without its module name, and all other functions would NOT be imported.

```python
from numpy import arange
arange(5)
```
