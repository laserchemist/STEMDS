## 📝 Quick Review Quiz

````python
# Interactive quiz — run this cell and answer the questions!
import json
from IPython.display import display, HTML
import ipywidgets as widgets

with open('questions_02.json') as f:
    _quiz_questions = json.load(f)

def _run_quiz(questions):
    score_box = widgets.Output()
    all_feedback = []
    
    for qi, q in enumerate(questions):
        out = widgets.Output()
        label = widgets.HTML(f"<div class='question-box'><b>Q{qi+1}. {q['question']}</b></div>")
        opts = [a['answer'] for a in q['answers']]
        radio = widgets.RadioButtons(options=opts, layout=widgets.Layout(width='100%'))
        btn   = widgets.Button(description='Check Answer', button_style='primary',
                               layout=widgets.Layout(width='160px'))
        fb    = widgets.Output()
        
        def _check(b, _q=q, _radio=radio, _fb=fb, _btn=btn):
            _btn.disabled = True
            with _fb:
                for a in _q['answers']:
                    if a['answer'] == _radio.value:
                        icon = '✅' if a['correct'] else '❌'
                        display(HTML(f"<p>{icon} <i>{a['feedback']}</i></p>"))
                        break
        btn.on_click(_check)
        display(widgets.VBox([label, radio, btn, fb]))

_run_quiz(_quiz_questions)
````
