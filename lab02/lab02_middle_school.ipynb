## ⭐ Level 1 — Numbers: `int` and `float`

Python stores numbers in two main ways:

| Type | What it is | Example |
|------|-----------|---------|
| `int` | A **whole** number — no decimal | `13`, `-5`, `1000` |
| `float` | A number **with a decimal** point | `3.14`, `98.6`, `-0.5` |

Think of `int` as counting (you can't have 2.5 people!) and `float` as measuring (temperature, distance).

````python
# Run this to see types in action!
your_age   = 13          # whole number → int
height_m   = 1.62        # measurement → float
school     = "Lincoln"   # words → str  (we'll use this soon!)

print("Your age is a:", type(your_age))
print("Height is a: ", type(height_m))
print("School is a: ", type(school))
````

<div style="background:#c1fcc0;border-left:6px solid #90EE90;padding:12px 16px;border-radius:6px;margin:10px 0;">

### List/Array Manipulation
One of the most common things you'll do with lists/arrays is to modify each element in it using the same function or operation. How you modify differs between lists and arrays.

You can do math operations on an array, and Python will automatically apply that to all elements in the array. <br>(This is called element-wise operation.)

````python
# Vectorised arithmetic — no loop needed
temps_K = make_array(250, 275, 300, 325, 350)          # example array
temps_C = temps_K - 273.15                             # array - 273.15 (all terms are subtracted by 273.15)
print("K:", temps_K)
print("C:", np.round(temps_C, 2))
````

With lists, element manipulation is a lot more complicated, but also a lot more versatile.

<div style="background:#c1fcc0;border-left:6px solid #90EE90;padding:12px 16px;border-radius:6px;margin:10px 0;">
<b>New Structure!</b> <code><b>newList</b> = [<b>anything</b> for <b>x</b> in <b>listOrArray</b>]</code><br>
This is called List Comprehension. <code>x</code> represents each element in the list or array, and so any code involving <code>x</code> can be placed in the <code>anything</code> section. Several examples are given below.
</div>

````python
# example 1 (same K to C conversion as earlier)

list_temps_K = list(temps_K)                           # example list

list_temps_C = [x - 273.15 for x in list_temps_K]      # list comprehension (each term is subtracted by 273.15)

print("K:", list_temps_K)
print("C:", np.round(list_temps_C, 2))
````

````python
# example 2 (word lengths)

list_of_words = ['python', 'is', 'cool', 'but', 'sometimes', 'hard', 'to', 'work', 'with',
                 'and', 'that', 'gives', 'it', 'personality']
                                                                 # you can replace x with any other variable
ing_words = [len(word) for word in list_of_words]                # but you must use that variable to represent each term

[print(hi) for hi in list_of_words]                              # yet another example here (aint that cool?)
print("Word lengths:", ing_words)
````

### `np.arange` — Mechanics & Edge Cases
To easily generate numbers in a certain sequence, use `np.arange`.
<div style="background:#c1fcc0;border-left:6px solid #90EE90;padding:12px 16px;border-radius:6px;margin:10px 0;">
<b>New Function!</b> <code>np.arange(<b>start</b>, <b>stop</b>, <b>step</b>)</code><br>
This generates numbers starting from <code>start</code> and increasing by <code>step</code> until it reaches the <code>end</code>.<br>
You can omit <code>step</code> from your code; in that case, python assumes the step is 1.<br>
Likewise, you can omit <code>start</code> from your code; in that case, python assumes the start is 0.
</div>

````python
# Verify the pattern with small examples:

print("Multiples of 3 up to 12:", np.arange(0, 13, 3))   # stop = 12 + 1
print("Multiples of 7 up to 35:", np.arange(0, 36, 7))   # stop = 35 + 1
````
