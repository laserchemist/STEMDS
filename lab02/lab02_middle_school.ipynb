## ⭐ Level 1 — Numbers: `int` and `float`

Python stores numbers in two main ways:

| Type | What it is | Example |
|------|-----------|---------|
| `int` | A **whole** number — no decimal | `13`, `-5`, `1000` |
| `float` | A number **with a decimal** point | `3.14`, `98.6`, `-0.5` |

Think of `int` as counting (you can't have 2.5 people!) and `float` as measuring (temperature, distance).

````python
# Run this to see types in action!
your_age   = 13          # whole number → int
height_m   = 1.62        # measurement → float
school     = "Lincoln"   # words → str  (we'll use this soon!)

print("Your age is a:", type(your_age))
print("Height is a: ", type(height_m))
print("School is a: ", type(school))
````

## ⭐⭐⭐ Level 4 — String Methods: Text Superpowers 🦸

Methods are like functions that belong **to** a string.  
You call them with a dot: `string.method()`

| Method | What it does | Example |
|--------|-------------|---------|
| `.upper()` | ALL CAPS | `"hello".upper()` → `"HELLO"` |
| `.lower()` | all lowercase | `"HI".lower()` → `"hi"` |
| `.replace(old, new)` | swap text | `"cat".replace("c","b")` → `"bat"` |
| `.len()` | nope — use `len(string)` | `len("hi")` → `2` |

````python
# String methods in action!
cheer = "go owls!"
print(cheer.upper())                        # SHOUT IT!
print(cheer.title())                        # Title Case
print(cheer.replace("owls", "python"))      # swap a word
````

## ⭐⭐⭐⭐ Level 5 — Arrays: Power in Numbers! 💪

An **array** is like a row of labelled boxes that all hold the same type of value.  
The real magic: you can do math on **all values at once** — no loop needed!

```
temperatures = make_array(18.0, 21.5, 19.0)
temperatures + 5   →  array([23.0, 26.5, 24.0])   # adds 5 to EVERY element!
```

Two ways to build arrays:

| Function | What it does |
|----------|-------------|
| `make_array(v1, v2, v3)` | You list each value by hand |
| `np.arange(start, stop, step)` | Automatically fills a range |

````python
# make_array example
temps_C = make_array(0, 20, 37, 100)    # freezing, room, body, boiling
temps_F = temps_C * 9/5 + 32            # convert ALL at once!
print("°C:", temps_C)
print("°F:", temps_F)
````

````python
# np.arange: count up automatically
# np.arange(start, stop, step) — stops BEFORE 'stop'
countdown = np.arange(10, 0, -1)   # 10 down to 1
evens      = np.arange(0, 11, 2)   # 0, 2, 4, 6, 8, 10
print("Countdown:", countdown)
print("Evens 0-10:", evens)
````

### 🔍 Understanding `np.arange` — Step by Step

`np.arange(start, stop, step)` works exactly like counting:

```
np.arange(1, 6, 1)  →  [1, 2, 3, 4, 5]     ← stops BEFORE 6
np.arange(0, 10, 2) →  [0, 2, 4, 6, 8]     ← steps of 2, stops before 10
np.arange(5, 0, -1) →  [5, 4, 3, 2, 1]     ← counts DOWN
```

**The stop value is never included.** Think of it like a slice of pizza —  
you stop cutting *before* you get to the edge!

🧪 **Try it yourself** — predict the output before you run:

````python
# Experiment: predict before running each line!
print(np.arange(1, 5))          # what values will this give?
print(np.arange(0, 20, 5))      # what about this one?
print(np.arange(100, 0, -25))   # and this countdown?
````

### 🎮 Boss Level: Multiples!

Before the big challenge, let's practise with `np.arange` for multiples.

A **multiple** of a number is what you get when you multiply it by 0, 1, 2, 3, …

```
Multiples of 5:   0, 5, 10, 15, 20, 25, ...   (step = 5)
Multiples of 10:  0, 10, 20, 30, ...           (step = 10)
Multiples of 99:  0, 99, 198, 297, ...         (step = 99)
```

With `np.arange(start, stop, step)`, the **step** IS the number you're multiplying by!

````python
# Warm-up: multiples of 5 from 0 to 25 (including 25!)
multiples_of_5 = np.arange(0, 26, 5)
print("Multiples of 5:", multiples_of_5)
````
