## 1. Data Types

Every value in Python has a **type** that governs valid operations and storage.

```python
type(42)        # → <class 'int'>
```

````python
print(int(3.99))        # 3   — truncated
print(float(7))         # 7.0
print(str(42))          # '42'
print(int("15"))        # 15
# print(int("3.14"))    # ← TypeError: why?
````

## 6. Lists & Arrays

| Feature | Python `list` | NumPy `array` |
|---------|--------------|---------------|
| Mixed types | ✅ | ❌ |
| Element-wise math | ❌ (need loop) | ✅ direct |
| Performance on large data | Slow | Fast |

````python
# Vectorised arithmetic — no loop needed
temps_K = make_array(250, 275, 300, 325, 350)
temps_C = temps_K - 273.15
print("K:", temps_K)
print("C:", np.round(temps_C, 2))
````

````python
# the same operation with a list
list_temps_K = list(temps_K)
list_temps_C = [x - 273.15 for x in list_temps_K]
# this is List Comprehension;
# general structure is [<any code involving x> for x in <list or array>]
# you can substitute x out for any variable, but both x's must be the same.
print("K:", list_temps_K)
print("C:", np.round(list_temps_C, 2))
````

### `np.arange` — Mechanics & Edge Cases

`np.arange(start, stop, step)` generates values in `[start, stop)` — the stop is **exclusive**.

```python
np.arange(0, 10, 3)   →  [0, 3, 6, 9]      # stops before 10
np.arange(1, 6)        →  [1, 2, 3, 4, 5]   # default step = 1
np.arange(0, 1, 0.25)  →  [0, 0.25, 0.5, 0.75]
```

**Key insight for multiples:** to generate multiples of *n* up to and *including* a target *T*,  
the stop must be `T + n` (one step past the last value you want).

````python
# Verify the pattern with small examples:
print("Multiples of 3 up to 12:", np.arange(0, 15, 3))   # stop = 12 + 3
print("Multiples of 7 up to 35:", np.arange(0, 42, 7))   # stop = 35 + 7
````
